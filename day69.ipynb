```python
import tensorflow as tf
import numpy as np
```

```python
num_samples = 1000
num_features = 10
task_a_labels = np.random.randint(0, 2, size=(num_samples,))  # Binary labels for task A
task_b_labels = np.random.randint(0, 3, size=(num_samples,))  # Categorical labels for task B (3 classes)
```

```python
x_train = np.random.random((num_samples, num_features)).astype(np.float32)
```

```python
def create_multi_task_model():
    inputs = tf.keras.Input(shape=(num_features,))

    # Shared layers
    shared_layer = tf.keras.layers.Dense(64, activation='relu')(inputs)
    shared_layer = tf.keras.layers.Dense(32, activation='relu')(shared_layer)

    # Task A output
    task_a_output = tf.keras.layers.Dense(1, activation='sigmoid', name='task_a')(shared_layer)

    # Task B output
    task_b_output = tf.keras.layers.Dense(3, activation='softmax', name='task_b')(shared_layer)

    model = tf.keras.Model(inputs=inputs, outputs=[task_a_output, task_b_output])
    return model
```

```python
model = create_multi_task_model()
model.compile(optimizer='adam',
              loss={'task_a': 'binary_crossentropy', 'task_b': 'sparse_categorical_crossentropy'},
              metrics={'task_a': 'accuracy', 'task_b': 'accuracy'})
```

```python
model.fit(x_train, {'task_a': task_a_labels, 'task_b': task_b_labels}, epochs=10, batch_size=32)
```

```python
loss, task_a_loss, task_b_loss, task_a_accuracy, task_b_accuracy = model.evaluate(x_train,
    {'task_a': task_a_labels, 'task_b': task_b_labels})
```

```python
print(f'Task A Loss: {task_a_loss:.4f}, Task A Accuracy: {task_a_accuracy:.4f}')
print(f'Task B Loss: {task_b_loss:.4f}, Task B Accuracy: {task_b_accuracy:.4f}')
```
